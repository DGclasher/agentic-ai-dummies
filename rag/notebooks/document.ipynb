### Loading documents

```python
from langchain_core.documents import Document
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_texts = {
    "../data/text_files/python.txt": """
        Python is a high-level, interpreted, general-purpose programming language known for its simplicity, readability, and versatility. It was created by Guido van Rossum and first released in 1991. Python uses a clean and straightforward syntax that allows developers to write programs with fewer lines of code compared to many other programming languages. This makes it especially popular among beginners, while its extensive ecosystem also makes it suitable for professional software development. Python supports multiple programming paradigms, including procedural, object-oriented, and functional programming. It is widely used in web development, data science, machine learning, artificial intelligence, automation, scripting, cybersecurity, scientific computing, and application development. Popular frameworks and libraries such as Django, Flask, FastAPI, NumPy, Pandas, TensorFlow, PyTorch, and Scikit-learn have contributed significantly to its popularity. Python also provides a powerful package manager and a huge collection of third-party packages through the Python Package Index (PyPI), allowing developers to easily add functionality to their applications. Another important feature of Python is its cross-platform compatibility, meaning that Python programs can generally run on Windows, Linux, and macOS with little or no modification. Python has strong community support, with extensive documentation, tutorials, forums, and open-source projects available to developers. It is also commonly used for automation because developers can quickly create scripts for repetitive tasks such as file manipulation, system administration, data processing, and API interactions. Python's dynamic typing allows variables to be created without explicitly declaring their types, making development faster, although developers need to be careful about type-related errors. Python also includes useful features such as exception handling, modules, packages, generators, decorators, comprehensions, and asynchronous programming. Despite being slower than compiled languages such as C++ or Rust for certain workloads, Python remains extremely popular because development speed and ecosystem support are often more important than raw execution speed. Overall, Python is a powerful and flexible language that can be used for everything from small automation scripts to large-scale production systems.""",
}

for filepath, content in sample_texts.items():
    with open(filepath, 'w') as f:
        f.write(content)
```

```python
## Textloaders
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python.txt", encoding="utf-8")
txt_doc = loader.load()
txt_doc
```

```python
# Directory loader
from langchain_community.document_loaders import DirectoryLoader

loader = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
)

dir_doc = loader.load()
dir_doc
```

```python
from langchain_community.document_loaders import PyMuPDFLoader

loader = DirectoryLoader(
    "../data/pdf_files",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=False
)

doc_pdf = loader.load()
doc_pdf
```
