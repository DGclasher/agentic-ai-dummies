```python
import os
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["OPENROUTER_API_KEY"] = os.getenv("OPENROUTER_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(model_provider="groq", model="llama-3.3-70b-versatile")
model
```

```python
response = model.invoke("Hello, who are you?")
response
```

### Streaming and Batch

Models stream their output while it is being generated, by displaying the output progressively we can enhance the UX particularly for longer responses.

```python
response = model.invoke("Write me a 200 words para about artificial intelligence and its impact on society.")
display(Markdown(response.content))
```

```python
for chunk in model.stream("Write me a 200 words para about artificial intelligence and its impact on society."):
    print(chunk.text, end="", flush=True)
    
```

```python
## In markdown format

response = ""
display_handle = display(Markdown(response), display_id=True)

for chunk in model.stream("Write me a 200 words para about artificial intelligence and its impact on society."):
    response += chunk.text
    display_handle.update(Markdown(response))
```

### Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce cost as the processing can be done parallely.

```python
responses = model.batch([
    "How do airplanes fly? explain in short",
    "What are the seven colors in a rainbow?",
    "How many genders are there in the world? answer in short"
])

for response in responses:
    display(Markdown(response.content))
```

```python
responses = model.batch([
    "How do airplanes fly? explain in short",
    "What are the seven colors in a rainbow?",
    "How many genders are there in the world? answer in short"
],
config={
    "max_concurrency": 5, # This limits this to 5 concurrent requests at a time
}
)
```
