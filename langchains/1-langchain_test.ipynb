```python
from dotenv import load_dotenv
import os

load_dotenv()
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
import requests

def get_weather(city:str) -> str:
    """This function returns the weather for a given city."""
    ENDPOINT = f"https://wttr.in/{city}?format=j1"
    # Implementation for fetching weather data would go here
    response = requests.get(ENDPOINT)
    weather_data = response.json()
    return f"The weather in {city} is currently {weather_data['current_condition'][0]['weatherDesc'][0]['value']}."

# get_weather("Kolkata")
```

```python
## Simple chat interface
```

```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq


llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=GROQ_API_KEY
)

agent = create_agent(
    model=llm,
    tools=[get_weather],
    system_prompt="You are a helpful assistant.",
)
# agent
response = agent.invoke({"messages": [{"role": "user", "content": "What is the weather in Bengaluru?"}]})
```
